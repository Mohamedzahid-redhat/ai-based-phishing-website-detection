```python

from datetime import datetime, timezone
import whois

def get_domain_registration_length(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        from urllib.parse import urlparse
        domain = urlparse(url).hostname

        info = whois.whois(domain)
        expiry = info.expiration_date

        if isinstance(expiry, list):
            expiry = expiry[0]

        if expiry is None:
            return None

        if expiry.tzinfo is None:
            expiry = expiry.replace(tzinfo=timezone.utc)

        days_left = (expiry - datetime.now(timezone.utc)).days

        if days_left > 365:
            return 1
        else:
            return -1

    except Exception as e:
        print("WHOIS lookup unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_favicon_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")

        icon_link = soup.find(
            "link",
            rel=lambda value: value and "icon" in str(value).lower()
        )

        if icon_link and icon_link.get("href"):
            favicon_url = urljoin(url, icon_link["href"])
        else:
            favicon_url = urljoin(url, "/favicon.ico")

        website_domain = urlparse(url).hostname
        favicon_domain = urlparse(favicon_url).hostname

        if website_domain == favicon_domain:
            return 1
        else:
            return -1

    except Exception as e:
        print("Favicon extraction unavailable:", e)
        return None
```

```python

from urllib.parse import urlparse

def extract_port_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        parsed = urlparse(url)
        port = parsed.port

        if port is None:
            return 1

        if port in [80, 443]:
            return 1
        else:
            return -1

    except Exception as e:
        print("Port extraction error:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_request_url_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        website_domain = urlparse(url).hostname

        resources = []

        for tag in soup.find_all(["img", "audio", "video", "script", "link"]):
            resource = (
                tag.get("src") or
                tag.get("href") or
                tag.get("data-src")
            )

            if resource:
                resources.append(urljoin(url, resource))

        if not resources:
            return None

        external_count = sum(
            1 for resource in resources
            if urlparse(resource).hostname != website_domain
        )

        external_percentage = (external_count / len(resources)) * 100

        if external_percentage < 22:
            return 1
        else:
            return -1

    except Exception as e:
        print("Request URL extraction unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_url_of_anchor_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        website_domain = urlparse(url).hostname

        anchors = soup.find_all("a", href=True)

        if not anchors:
            return None

        external_count = 0

        for anchor in anchors:
            link = urljoin(url, anchor["href"])
            link_domain = urlparse(link).hostname

            if link_domain and link_domain != website_domain:
                external_count += 1

        external_percentage = (external_count / len(anchors)) * 100

        if external_percentage < 31:
            return 1
        else:
            return -1

    except Exception as e:
        print("Anchor URL extraction unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_sfh_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        website_domain = urlparse(url).hostname

        forms = soup.find_all("form")

        if not forms:
            return None

        for form in forms:
            action = form.get("action", "").strip()

            if not action:
                return -1

            destination = urljoin(url, action)
            destination_domain = urlparse(destination).hostname

            if destination_domain != website_domain:
                return -1

        return 1

    except Exception as e:
        print("SFH extraction unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup
import re

def extract_rightclick_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")

        # Check inline JavaScript
        page_source = response.text.lower()

        if re.search(r"event\.button\s*==\s*2", page_source):
            return -1

        # Check oncontextmenu handlers
        for tag in soup.find_all(oncontextmenu=True):
            script = str(tag.get("oncontextmenu", "")).lower()

            if "return false" in script:
                return -1

        return 1

    except Exception as e:
        print("Right-click extraction unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup

def extract_iframe_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")

        iframes = soup.find_all("iframe")

        for iframe in iframes:
            width = iframe.get("width", "")
            height = iframe.get("height", "")
            style = iframe.get("style", "").lower()

            # Check for hidden or invisible iframe
            if (
                width == "0"
                or height == "0"
                or "display:none" in style.replace(" ", "")
                or "visibility:hidden" in style.replace(" ", "")
            ):
                return -1

        return 1

    except Exception as e:
        print("IFrame feature extraction unavailable:", e)
        return None
```

```python

from datetime import datetime, timezone
from urllib.parse import urlparse
import whois

def extract_domain_age_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        domain = urlparse(url).hostname

        info = whois.whois(domain)
        creation_date = info.creation_date

        if isinstance(creation_date, list):
            creation_date = creation_date[0]

        if creation_date is None:
            print("Domain creation date unavailable")
            return None

        if creation_date.tzinfo is None:
            creation_date = creation_date.replace(
                tzinfo=timezone.utc
            )

        age_days = (
            datetime.now(timezone.utc) - creation_date
        ).days

        print("Domain:", domain)
        print("Domain age in days:", age_days)

        if age_days >= 180:
            return 1
        else:
            return -1

    except Exception as e:
        print("Domain age lookup unavailable:", e)
        return None
```

```python
import requests
from urllib.parse import urlparse

def extract_ssl_final_state(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=10,
            allow_redirects=True,
            headers={"User-Agent": "Mozilla/5.0"},
            verify=True
        )

        if urlparse(response.url).scheme.lower() == "https":
            print("HTTPS certificate verification successful.")
            return 1
        else:
            print("Final webpage does not use HTTPS.")
            return -1

    except requests.exceptions.SSLError:
        print("SSL certificate verification failed.")
        return -1

    except Exception as e:
        print("SSL verification unavailable:", e)
        return None
```

```python
import whois
from datetime import datetime, timezone
from urllib.parse import urlparse

def get_domain_registration_length(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        domain = urlparse(url).hostname
        info = whois.whois(domain)

        expiry = info.expiration_date

        if isinstance(expiry, list):
            expiry = expiry[0]

        if expiry is None:
            return None

        if expiry.tzinfo is None:
            expiry = expiry.replace(tzinfo=timezone.utc)

        days_left = (
            expiry - datetime.now(timezone.utc)
        ).days

        print("Domain:", domain)
        print("Days until expiry:", days_left)

        return 1 if days_left > 365 else -1

    except Exception as e:
        print("WHOIS lookup unavailable:", e)
        return None
```

```python

import requests
import re
import tldextract
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_url_of_anchor_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        anchors = soup.find_all("a")

        if not anchors:
            print("No anchor links found.")
            return None

        domain_extractor = tldextract.TLDExtract(
            suffix_list_urls=()
        )

        page_host = urlparse(response.url).hostname or ""
        page_domain = (
            domain_extractor(page_host).top_domain_under_public_suffix
            or page_host
        )

        suspicious_count = 0

        for anchor in anchors:
            href = (anchor.get("href") or "").strip()

            # Links with no real webpage destination
            if (
                not href
                or href.startswith("#")
                or re.match(r"^\s*javascript\s*:", href, re.I)
            ):
                suspicious_count += 1
                continue

            target_url = urljoin(response.url, href)
            parsed = urlparse(target_url)

            if parsed.scheme not in ("http", "https"):
                suspicious_count += 1
                continue

            target_host = parsed.hostname or ""
            target_domain = (
                domain_extractor(target_host).top_domain_under_public_suffix
                or target_host
            )

            if target_domain != page_domain:
                suspicious_count += 1

        percentage = suspicious_count / len(anchors) * 100

        print("Total anchor links:", len(anchors))
        print("Suspicious anchor links:", suspicious_count)
        print("Suspicious percentage:", round(percentage, 2), "%")

        # Simplified demo threshold, not an exact source rule
        return 1 if percentage < 31 else -1

    except Exception as e:
        print("URL of Anchor extraction unavailable:", e)
        return None
```

```python
result = extract_url_of_anchor_feature(
    "https://www.google.com"
)

print("URL of Anchor feature:", result)
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_links_in_tags_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        page_domain = urlparse(response.url).hostname

        resources = []

        for tag in soup.find_all(["meta", "script", "link"]):
            for attribute in ["content", "src", "href"]:
                value = tag.get(attribute)

                if value:
                    resource_url = urljoin(response.url, value)
                    parsed = urlparse(resource_url)

                    if parsed.scheme in ("http", "https"):
                        resources.append(parsed.hostname)

        if not resources:
            print("No URL resources found in these tags.")
            return None

        external_count = sum(
            domain is not None and domain != page_domain
            for domain in resources
        )

        percentage = external_count / len(resources) * 100

        print("Total tag resources:", len(resources))
        print("External resources:", external_count)
        print("External percentage:", round(percentage, 2), "%")

        # Simplified project threshold
        return 1 if percentage < 17 else -1

    except Exception as e:
        print("Links in Tags extraction unavailable:", e)
        return None
```

```python

import requests
import tldextract
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin

def extract_sfh_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")

        page_host = urlparse(response.url).hostname or ""
        page_domain = tldextract.extract(
            page_host
        ).top_domain_under_public_suffix or page_host

        forms = soup.find_all("form")

        if not forms:
            print("No forms found on the webpage.")
            return 1

        for form in forms:
            action = (form.get("action") or "").strip()

            if not action or action.lower() == "about:blank":
                print("Suspicious form action:", action)
                return -1

            target_url = urljoin(response.url, action)
            target_host = urlparse(target_url).hostname or ""

            target_domain = tldextract.extract(
                target_host
            ).top_domain_under_public_suffix or target_host

            if target_domain != page_domain:
                print("External form destination:", target_host)
                return -1

        print("Form actions point to the same domain.")
        return 1

    except Exception as e:
        print("SFH extraction unavailable:", e)
        return None
```

```python

import requests
import re
from bs4 import BeautifulSoup

def extract_popup_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        page_source = response.text.lower()

        popup_found = bool(
            re.search(r"window\.open\s*\(", page_source, re.I)
        )

        if not popup_found:
            print("No popup window code detected.")
            return 1

        personal_info_found = bool(
            re.search(
                r"password|credit.?card|personal.?information|"
                r"email.?address|social.?security",
                page_source,
                re.I
            )
        )

        if personal_info_found:
            print("Popup and possible personal-information request detected.")
            return -1

        print("Popup detected, but personal information request is unverified.")
        return None

    except Exception as e:
        print("Popup extraction unavailable:", e)
        return None
```

```python

import requests
from bs4 import BeautifulSoup

def extract_iframe_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        soup = BeautifulSoup(response.text, "html.parser")
        iframes = soup.find_all("iframe")

        for iframe in iframes:
            width = str(iframe.get("width", "")).strip()
            height = str(iframe.get("height", "")).strip()
            style = iframe.get("style", "").lower().replace(" ", "")

            if (
                width == "0"
                or height == "0"
                or "display:none" in style
                or "visibility:hidden" in style
            ):
                print("Hidden iframe detected.")
                return -1

        print("No hidden iframe detected.")
        return 1

    except Exception as e:
        print("IFrame extraction unavailable:", e)
        return None
```

```python

import whois
from datetime import datetime, timezone
from urllib.parse import urlparse

def extract_domain_age_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        domain = urlparse(url).hostname
        info = whois.whois(domain)

        creation_date = info.creation_date

        if isinstance(creation_date, list):
            creation_date = creation_date[0]

        if creation_date is None:
            print("Domain creation date unavailable.")
            return None

        if not isinstance(creation_date, datetime):
            print("WHOIS returned an unsupported date format.")
            return None

        if creation_date.tzinfo is None:
            creation_date = creation_date.replace(
                tzinfo=timezone.utc
            )

        age_days = (
            datetime.now(timezone.utc) - creation_date
        ).days

        print("Domain:", domain)
        print("Domain age in days:", age_days)

        return 1 if age_days >= 180 else -1

    except Exception as e:
        print("Domain age lookup unavailable:", e)
        return None
```

```python
result = extract_domain_age_feature(
    "https://www.google.com"
)

print("Age of Domain feature:", result)
```

```python

import dns.resolver
from urllib.parse import urlparse

def extract_dns_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        domain = urlparse(url).hostname

        dns.resolver.resolve(domain, "A")

        print("Domain:", domain)
        print("DNS record found!")
        return 1

    except dns.resolver.NXDOMAIN:
        print("Domain does not exist.")
        return -1

    except Exception as e:
        print("DNS lookup unavailable:", e)
        return None
```

```python

import requests
import tldextract
from urllib.parse import urlparse

def extract_web_traffic_feature(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        hostname = urlparse(url).hostname

        domain = tldextract.extract(
            hostname
        ).top_domain_under_public_suffix

        if not domain:
            return None

        api_url = (
            f"https://tranco-list.eu/api/ranks/domain/{domain}"
        )

        response = requests.get(api_url, timeout=15)
        response.raise_for_status()

        data = response.json()
        ranks = data.get("ranks", [])

        if not ranks:
            print("No Tranco ranking available.")
            return None

        latest = ranks[-1]
        rank = latest.get("rank")

        if rank is None:
            print("Ranking value unavailable.")
            return None

        print("Domain:", domain)
        print("Tranco rank:", rank)

        if rank <= 100000:
            return 1
        else:
            return 0

    except Exception as e:
        print("Web traffic lookup unavailable:", e)
        return None
```

```python

import requests
import tldextract

from datetime import datetime, timezone
from urllib.parse import urlparse

# Cache domain information to avoid repeated requests
rdap_cache = {}

def get_rdap_registration_dates(url):
    try:
        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        hostname = urlparse(url).hostname

        if not hostname:
            return None

        extracted = tldextract.extract(hostname)
        domain = extracted.top_domain_under_public_suffix

        if not domain:
            return None

        if domain in rdap_cache:
            return rdap_cache[domain]

        api_url = f"https://rdap.org/domain/{domain}"

        response = requests.get(api_url, timeout=20)
        response.raise_for_status()

        data = response.json()

        registration_date = None
        expiration_date = None

        for event in data.get("events", []):
            action = event.get("eventAction", "").lower()
            date_text = event.get("eventDate")

            if not date_text:
                continue

            date = datetime.fromisoformat(
                date_text.replace("Z", "+00:00")
            )

            if date.tzinfo is None:
                date = date.replace(tzinfo=timezone.utc)

            if action == "registration":
                registration_date = date

            elif action == "expiration":
                expiration_date = date

        result = {
            "domain": domain,
            "registration": registration_date,
            "expiration": expiration_date
        }

        rdap_cache[domain] = result

        return result

    except Exception as e:
        print("RDAP lookup unavailable:", e)
        return None


# Feature 9: Domain Registration Length
def get_domain_registration_length(url):

    info = get_rdap_registration_dates(url)

    if not info or not info["expiration"]:
        print("Domain expiry date unavailable.")
        return None

    days_left = (
        info["expiration"] - datetime.now(timezone.utc)
    ).days

    print("Domain:", info["domain"])
    print("Days until expiry:", days_left)

    return 1 if days_left > 365 else -1


# Feature 24: Age of Domain
def extract_domain_age_feature(url):

    info = get_rdap_registration_dates(url)

    if not info or not info["registration"]:
        print("Domain registration date unavailable.")
        return None

    age_days = (
        datetime.now(timezone.utc) - info["registration"]
    ).days

    print("Domain:", info["domain"])
    print("Domain age in days:", age_days)

    return 1 if age_days >= 180 else -1


print("RDAP registration functions created successfully!")
```
